# Part 1 – Load the Data

```python
# Load the three coffee datasets into separate DataFrames
exports = pd.read_csv("Coffee_export-1.csv")
imports = pd.read_csv("Coffee_import-1.csv")
production = pd.read_csv("Coffee_production-1.csv")
```

# Part 2 – Explore the Data

```python
# Group the three DataFrames so each question can be answered for all of them in one loop
datasets = {
    "Export": exports,
    "Import": imports,
    "Production": production,
}
```

```python
# Q1: How many rows and columns does each dataset contain?
for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
# Q2: What are the column names?
for name, df in datasets.items():
    print(f"--- {name} ---")
    print(list(df.columns))
    print()
```

```python
# Q3: What data types are assigned to each column?
for name, df in datasets.items():
    print(f"========== {name} ==========")
    df.info()
    print()
```

```python
# Q4: Are there any missing values?
for name, df in datasets.items():
    missing = df.isnull().sum()
    print(f"{name}: {missing.sum()} missing values in total")
    # Show only the columns that actually have missing values
    if missing.sum() > 0:
        print(missing[missing > 0])
    print()
```

```python
# Q5: Which span of years is represented?
# Years are stored as column names (wide format), so find columns that start with 4 digits
for name, df in datasets.items():
    year_cols = [col for col in df.columns if str(col)[:4].isdigit()]
    print(f"{name}: {year_cols[0]} to {year_cols[-1]} ({len(year_cols)} year columns)")
```

```python
# Q6: Which countries are included in the records?
country_col = "Country"  # change this if your column name is different

for name, df in datasets.items():
    countries = df[country_col].unique()
    print(f"--- {name}: {df[country_col].nunique()} countries ---")
    print(sorted(countries))

    print()
```

**Notes**
- All three datasets are in **wide format**: each year is its own column instead of one `Year` column.
- The export and production files list producing countries (e.g., Brazil, Colombia, Ethiopia, Viet Nam). The import file lists consuming countries, mostly in Europe plus the United States of America.
- Production uses **crop years** (`1990/91`) and stores numbers as `float64`, while export and import use **calendar years** and `int64`. Comparing them directly will need some alignment.
- No values are missing, but some cells are `0`. These may hide missing data, which I will check in Part 4.
- In the import data, 28 country names (the EU members, e.g. `'   Austria'`) start with extra spaces, while the other 7 do not. Filtering or matching by country name will fail unless the spaces are removed.

# Part 3 – Descriptive Statistics

```python
# Show large numbers with commas instead of scientific notation (e.g. 1.2e+09)
pd.set_option("display.float_format", "{:,.0f}".format)

# Descriptive statistics for all numerical columns
for name, df in datasets.items():
    print(f"========== {name} ==========")
    display(df.describe())
```

```python
# Key metric = the Total column (sum of all 30 years for each country)
for name, df in datasets.items():
    total_col = [col for col in df.columns if col.startswith("Total")][0]
    total = df[total_col]

    print(f"--- {name} ({total_col}) ---")
    print(f"Largest : {total.max():,.0f}  ({df.loc[total.idxmax(), 'Country'].strip()})")
    print(f"Smallest: {total.min():,.0f}  ({df.loc[total.idxmin(), 'Country'].strip()})")
    print(f"Average : {total.mean():,.0f}")
    print(f"Median  : {total.median():,.0f}")
    print()
```

```python
# Top 5 countries by total, to see how concentrated the market is
for name, df in datasets.items():
    total_col = [col for col in df.columns if col.startswith("Total")][0]
    print(f"--- {name}: Top 5 ---")
    display(df.nlargest(5, total_col)[["Country", total_col]])
```

**Key metric:** the `Total` column of each dataset (sum of all 30 years per country).

| Dataset | Largest | Smallest | Average (mean) | Median |
|---|---|---|---|---|
| Export | 33,807,709,056 (Brazil) | 1,500,000 (Equatorial Guinea) | 2,851,331,801 | 316,080,000 |
| Import | 42,507,660,000 (United States of America) | 29,700,000 (Malta) | 5,069,693,143 | 1,640,040,000 |
| Production | 75,082,980,000 (Brazil) | 1,080,000 (Equatorial Guinea) | 4,104,269,455 | 623,640,000 |

**Observations and anomalies**
- **Brazil's exports in 2014 and 2015 are -2,147,483,648.** Exports can't be negative. This number is the smallest possible 32-bit integer, a common error code, so the real values are missing. Because `Total_export` adds these two values, Brazil's total is also wrong (about 4.3 billion too low).
- **The mean is much higher than the median** in all three datasets (export mean is about 9× the median). A few very large countries pull the average up, so the median describes a "typical" country better.
- **The market is very concentrated.** The top 5 countries make up about 61% of exports, 63% of imports, and 66% of production. Brazil alone produced about one-third of all coffee.
- The same four countries lead both export and production: Brazil, Viet Nam, Colombia, and Indonesia. The biggest importers are the United States of America and Germany.
- The minimum value in many year columns is **0**, so some countries have years with no recorded trade or production.

# Part 4 – Data Quality Assessment

```python
# Check missing values, fully duplicated rows, and repeated country names
for name, df in datasets.items():
    print(f"{name}: missing = {df.isnull().sum().sum()}, "
          f"duplicate rows = {df.duplicated().sum()}, "
          f"repeated countries = {df['Country'].str.strip().duplicated().sum()}")
```

```python
# Count how many years are recorded as 0 for each country
for name, df in datasets.items():
    year_cols = [col for col in df.columns if col[:4].isdigit()]
    zero_counts = (df[year_cols] == 0).sum(axis=1)
    has_zeros = df.loc[zero_counts > 0, "Country"].str.strip()

    print(f"--- {name}: {(zero_counts > 0).sum()} countries have 0 values ---")
    print(pd.Series(zero_counts[zero_counts > 0].values, index=has_zeros)
            .sort_values(ascending=False).head(5))
    print()
```

```python
# 1) Extra spaces in country names
spaces = imports["Country"] != imports["Country"].str.strip()
print(f"Import country names with extra spaces: {spaces.sum()}")

# 2) Country groups that changed over time
print(imports[imports["Country"].str.contains("Belgium|Luxembourg")]
      [["Country", "1997", "1998", "1999", "2000"]])

# 3) Two labels that may mean the same coffee type
print(production["Coffee type"].value_counts())

# 4) Different number types and year formats across files
print(f"\nExport years: {exports.columns[1]} ({exports['1990'].dtype}) | "
      f"Production years: {production.columns[2]} ({production['1990/91'].dtype})")
```

### Part 4 Answers

**Missing values and duplicates**
None of the three datasets have missing (`NaN`) values, duplicate rows, or repeated countries. However, "no missing values" does not mean the data is complete, because missing data is hidden in other forms (below).

| # | Issue | Where | Evidence | How it could break or bias analysis |
|---|---|---|---|---|
| 1 | **Error code instead of real values** | Export: Brazil, 2014–2015 | Value is `-2,147,483,648` (smallest 32-bit integer) | Brazil is the biggest exporter, so any total, average, or trend that includes these years is badly wrong. `Total_export` for Brazil is already about 4.3 billion too low. |
| 2 | **Zeros that may mean "no data"** | All three files | Export: 11 countries (Congo has 23 zero years). Import: 10 countries. Production: 7 countries | Some zeros happen before a country existed, like Timor-Leste (production 0 until 2000/01) and Croatia, Slovakia, and the Baltic states (import 0 in the early 1990s). Treating these as real zeros pulls averages down and creates fake growth when the numbers "start." |
| 3 | **Extra spaces in country names** | Import: 28 EU countries | `'   Germany'` instead of `'Germany'` | Filtering (`== "Germany"`) returns nothing, and merging with the export or production data by country fails silently. |
| 4 | **Country groups that changed** | Import: Belgium, Luxembourg | `Belgium/Luxembourg` has data until 1998, then becomes 0. `Belgium` and `Luxembourg` start in 1999 | Looking at one row alone shows a fake crash or a fake jump. The three rows have to be combined to see the real trend. |
| 5 | **Inconsistent category labels** | Production: `Coffee type` | `Arabica/Robusta` (8) vs. `Robusta/Arabica` (10) | If they mean "mixed," a `groupby` will split one group into two and make each look smaller. If the order means "main type first," that rule should be documented. |
| 6 | **Different year formats and types** | Production vs. export and import | Crop years `1990/91` as `float64` vs. calendar years `1990` as `int64` | Columns can't be matched directly when comparing production to exports for the same year. A crop year also covers parts of two calendar years. |
| 7 | **Wide format** | All three files | 30 year columns instead of one `Year` column | Hard to plot trends over time or filter by year. The data will likely need to be reshaped to long format with `melt()`. |

# Part 5 – Initial Insights

Which dataset appears to contain the most complete information?
 -  The **Production Dataset** contains the most complete information. It also includes an extra Coffee type column which gives more detail than the other two files. 

Which dataset looks like it will require the most extensive cleaning?
 - The **Import Dataset** required the most extensive cleaning. 28 of its 35 country names start with extra spaces, Belgium and Luxembourg are split into three rows that overlap in different years, and several countries show zeros in the early 1990s simply because they did not exist yet. 
 - The **Export Dataset** has critical single error. Brazil, the largest exporter, has an error code (-2,147,483,648) instead of real values for 2014 and 2015.

What specific types of business questions could these datasets help a global coffee corporation answer?
