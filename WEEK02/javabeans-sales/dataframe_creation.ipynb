# JavaBeans Data Analysis
## Part 1 – Construct the DataFrame

```python
import pandas as pd
```

```python
# Store sales data in dictionary for "coffee_sales" DataFrame
#   key = drink(column), value = [2021 Sales, 2022 Sales]
sales_data = {
    "Americano": [5628, 6239],
    "Latte": [12426, 25384],
    "Espresso": [852, 756],
    "Cappuccino": [8522, 7561]
}

# Create the DataFrame, using pandas
coffee_sales = pd.DataFrame(sales_data, index=["2021 Sales", "2022 Sales"])
coffee_sales.index.name = "Year"

coffee_sales
```

```python
# Compare types: the whole table vs a single column
print(type(coffee_sales))
print(type(coffee_sales["Latte"]))
```

```python
# Verify the data type of each column
coffee_sales.dtypes
```

## Part 2 – Data Manipulation

### 1. Total Product Volume

```python
total_volume = coffee_sales.sum()

total_volume
```

### 2. Year-over-Year Growth

```python
sales_2021 = coffee_sales.loc["2021 Sales"]
sales_2022 = coffee_sales.loc["2022 Sales"]

yoy_growth = ((sales_2022 - sales_2021) / sales_2021 * 100).round(2)

yoy_growth
```

## Part 3 – Business Insights & Strategy

### 1. Identify the absolute best-selling and least-selling beverage for both 2021 and 2022

```python
best_seller = coffee_sales.idxmax(axis=1)
least_seller = coffee_sales.idxmin(axis=1)

print("Best seller:\n", best_seller)
print("\nLeast seller:\n", least_seller)
```

### 2. Strategic Analysis for JavaBeans

The Latte has most powerful growth and sales (grew by 104.28% in a year, from 12,426 to 25,384.) That means, Latte deserves higher marketing spend. Furthermore, making new drinks based on the latte such as Black sugar latte could be a strategy for growing more sales.

Espresso and Cappuccino declined, so we could consider removing them. However the volume of Cappuccino is way bigger than Espresso, so we can consider redesign of Cappuccino. However, this analysis is based on only two years of data. We should get and analyze more data.
